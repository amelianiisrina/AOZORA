# Membuat data transaksi

```python
# Menentukan distribusi waktu transaksi
def generate_order_hour():
    hour_ranges = [
        list(range(10, 11)),  # 10.00–10.59
        list(range(11, 14)),  # 11.00–13.59
        list(range(14, 18)),  # 14.00–17.59
        list(range(18, 22))   # 18.00–21.59
    ]

    selected_range = random.choices(
        hour_ranges,
        weights=[5, 30, 15, 50]
    )[0]

    return random.choice(selected_range)

# Membuat data transaksi
orders_data = []

branch_ids = branches["branch_id"].tolist()

for i in range(1, TOTAL_ORDERS + 1):

    order_date = random.choice(dates)
    order_hour = generate_order_hour()

    order_datetime = (
        pd.Timestamp(order_date)
        + pd.Timedelta(
            hours=order_hour,
            minutes=random.randint(0, 59)
        )
    )

    branch_id = random.choice(branch_ids)

    order_type = random.choices(
        ["Dine-in", "Takeaway", "Delivery"],
        weights=[50, 25, 25]
    )[0]

    status = random.choices(
        ["Completed", "Cancelled"],
        weights=[95, 5]
    )[0]

    prep_time = max(
        2,
        int(np.random.normal(14, 5))
    )

    orders_data.append({
        "order_id": f"O{i:04d}",
        "branch_id": branch_id,
        "order_datetime": order_datetime,
        "order_type": order_type,
        "status": status,
        "prep_time_minutes": prep_time
    })

orders = pd.DataFrame(orders_data)

display(orders.head(10))
```

# Menghitung pemakaian bahan berdasarkan penjualan

```python
# Mengambil transaksi yang berhasil diselesaikan
completed_orders = orders[
    orders["status"] == "Completed"
].copy()

# Mengubah tanggal transaksi menjadi format tanggal saja
completed_orders["record_date"] = pd.to_datetime(
    completed_orders["order_datetime"]
).dt.date

# Menggabungkan order_items dengan orders
sales_detail = order_items.merge(
    completed_orders[["order_id", "branch_id", "record_date"]],
    on="order_id",
    how="inner"
)

# Menggabungkan detail penjualan dengan komposisi resep
usage_detail = sales_detail.merge(
    menu_recipes[["menu_id", "item_id", "quantity_per_portion"]],
    on="menu_id",
    how="inner"
)

# Menghitung jumlah bahan yang digunakan
usage_detail["used_quantity"] = (
    usage_detail["quantity"] *
    usage_detail["quantity_per_portion"]
)

# Mengelompokkan pemakaian berdasarkan tanggal, cabang, dan bahan
daily_usage = (
    usage_detail
    .groupby(["record_date", "branch_id", "item_id"], as_index=False)
    ["used_quantity"]
    .sum()
)

display(daily_usage.head(10))

print("Jumlah catatan pemakaian bahan:", len(daily_usage))
print("Jumlah cabang:", daily_usage["branch_id"].nunique())
print("Jumlah bahan:", daily_usage["item_id"].nunique())
```

# Membuat data inventory harian

```python
inventory_data = []

inventory_counter = 1

# Membuat catatan untuk setiap cabang dan bahan
for branch_id in branches["branch_id"]:
    for _, ingredient in ingredients.iterrows():

        item_id = ingredient["item_id"]
        minimum_stock = float(ingredient["minimum_stock"])
        unit_cost = float(ingredient["unit_cost"])

        # Mengambil pemakaian bahan untuk cabang dan bahan tersebut
        item_usage = daily_usage[
            (daily_usage["branch_id"] == branch_id) &
            (daily_usage["item_id"] == item_id)
        ]

        # Menghitung rata-rata pemakaian harian
        average_usage = (
            item_usage["used_quantity"].mean()
            if len(item_usage) > 0 else 0
        )

        # Menentukan stok awal pada hari pertama
        opening_stock = max(
            minimum_stock * 2,
            minimum_stock + average_usage * 3
        )

        # Membuat catatan untuk setiap tanggal
        for date in dates:

            record_date = date.date()

            # Mengambil pemakaian bahan pada tanggal tersebut
            usage_match = item_usage[
                item_usage["record_date"] == record_date
            ]

            used_quantity = (
                float(usage_match["used_quantity"].iloc[0])
                if len(usage_match) > 0 else 0
            )

            # Menghitung bahan yang terbuang (simulasi 1-4%)
            waste_quantity = (
                used_quantity * random.uniform(0.01, 0.04)
            )

            # Menentukan stok masuk jika stok diperkirakan
            # turun di bawah batas minimum
            estimated_stock = (
                opening_stock - used_quantity - waste_quantity
            )

            target_stock = max(
                minimum_stock * 2,
                minimum_stock + average_usage * 3
            )

            received_quantity = (
                max(0, target_stock - estimated_stock)
                if estimated_stock < minimum_stock else 0
            )

            # Menghitung stok akhir
            closing_stock = (
                opening_stock
                + received_quantity
                - used_quantity
                - waste_quantity
            )

            inventory_data.append({
                "inventory_record_id": f"INV{inventory_counter:05d}",
                "record_date": record_date,
                "branch_id": branch_id,
                "item_id": item_id,
                "opening_stock": round(opening_stock, 2),
                "received_quantity": round(received_quantity, 2),
                "used_quantity": round(used_quantity, 2),
                "waste_quantity": round(waste_quantity, 2),
                "closing_stock": round(closing_stock, 2),
                "minimum_stock": minimum_stock,
                "unit_cost": unit_cost,
                "waste_cost": round(waste_quantity * unit_cost, 2),
                "used_cost": round(used_quantity * unit_cost, 2)
            })

            # Stok akhir hari ini menjadi stok awal hari berikutnya
            opening_stock = closing_stock

            inventory_counter += 1

inventory = pd.DataFrame(inventory_data)

display(inventory.head(10))

print("Jumlah baris inventory:", len(inventory))
print("Jumlah cabang:", inventory["branch_id"].nunique())
print("Jumlah bahan:", inventory["item_id"].nunique())
print("Jumlah tanggal:", inventory["record_date"].nunique())
```

# Membuat data table_sessions

```python
# Mengambil transaksi dine-in yang berhasil diselesaikan
dine_in_orders = orders[
    (orders["order_type"] == "Dine-in") &
    (orders["status"] == "Completed")
].copy()

dine_in_orders = dine_in_orders.sort_values(
    "order_datetime"
)

tables_by_branch = {
    branch_id: tables[
        tables["branch_id"] == branch_id
    ]["table_id"].tolist()
    for branch_id in branches["branch_id"]
}

table_available_at = {}
session_data = []

for i, (_, order) in enumerate(
    dine_in_orders.iterrows(), start=1
):

    branch_id = order["branch_id"]
    start_time = pd.Timestamp(order["order_datetime"])

    # Batas waktu operasional
    closing_time = start_time.normalize() + pd.Timedelta(hours=22)

    # Mencari meja yang tersedia
    available_tables = [
        table_id
        for table_id in tables_by_branch[branch_id]
        if table_available_at.get(
            table_id, pd.Timestamp.min
        ) <= start_time
    ]

    if available_tables:
        table_id = random.choice(available_tables)
    else:
        table_id = min(
            tables_by_branch[branch_id],
            key=lambda t: table_available_at.get(
                t, pd.Timestamp.min
            )
        )

        start_time = table_available_at[table_id]

    # Menghitung sisa waktu operasional
    remaining_minutes = int(
        (closing_time - start_time).total_seconds() / 60
    )

    # Lewati sesi jika tidak ada waktu tersisa
    if remaining_minutes <= 0:
        continue

    # Durasi maksimal mengikuti sisa waktu operasional
    duration_minutes = min(
        random.randint(20, 90),
        remaining_minutes
    )

    end_time = (
        start_time +
        pd.Timedelta(minutes=duration_minutes)
    )

    session_data.append({
        "session_id": f"S{len(session_data) + 1:05d}",
        "table_id": table_id,
        "order_id": order["order_id"],
        "start_time": start_time,
        "end_time": end_time,
        "duration_minutes": duration_minutes
    })

    table_available_at[table_id] = end_time

table_sessions = pd.DataFrame(session_data)

display(table_sessions.head(10))

print("Jumlah sesi meja:", len(table_sessions))
print("Jumlah transaksi unik:", table_sessions["order_id"].nunique())
print("Jumlah meja unik:", table_sessions["table_id"].nunique())
```

```python
# Pemeriksaan primary key
primary_keys = {
    "branches": ("branch_id", branches),
    "menu_cost": ("menu_id", menu_cost),
    "ingredients": ("item_id", ingredients),
    "menu_recipes": ("composition_id", menu_recipes),
    "tables": ("table_id", tables),
    "orders": ("order_id", orders),
    "order_items": ("order_item_id", order_items),
    "inventory": ("inventory_record_id", inventory),
    "table_sessions": ("session_id", table_sessions),
    "feedback": ("feedback_id", feedback)
}

print("=== PRIMARY KEY ===")

for name, (column, df) in primary_keys.items():
    print(
        name,
        ":", "Valid" if df[column].is_unique else "Duplikat ditemukan"
    )


# Pemeriksaan foreign key
foreign_keys = [
    ("orders", orders, "branch_id", branches, "branch_id"),
    ("tables", tables, "branch_id", branches, "branch_id"),
    ("menu_recipes", menu_recipes, "menu_id", menu_cost, "menu_id"),
    ("menu_recipes", menu_recipes, "item_id", ingredients, "item_id"),
    ("order_items", order_items, "order_id", orders, "order_id"),
    ("order_items", order_items, "menu_id", menu_cost, "menu_id"),
    ("inventory", inventory, "branch_id", branches, "branch_id"),
    ("inventory", inventory, "item_id", ingredients, "item_id"),
    ("table_sessions", table_sessions, "table_id", tables, "table_id"),
    ("table_sessions", table_sessions, "order_id", orders, "order_id"),
    ("feedback", feedback, "order_id", orders, "order_id")
]

print("\n=== FOREIGN KEY ===")

for child_name, child_df, child_col, parent_df, parent_col in foreign_keys:
    invalid = ~child_df[child_col].isin(parent_df[parent_col])

    print(
        f"{child_name}.{child_col}:",
        f"{invalid.sum()} relasi tidak valid"
    )
```

```python
# 1. Validasi waktu transaksi
order_hours_valid = (
    (orders["order_datetime"].dt.hour >= 10) &
    (orders["order_datetime"].dt.hour < 22)
).all()

# 2. Validasi subtotal order_items
subtotal_valid = np.allclose(
    order_items["subtotal"],
    order_items["quantity"] * order_items["unit_price"],
    atol=0.01
)

# 3. Validasi harga menu pada order_items
price_check = order_items.merge(
    menu_cost[["menu_id", "selling_price"]],
    on="menu_id",
    how="left"
)

price_valid = (
    price_check["unit_price"] ==
    price_check["selling_price"]
).all()

# 4. Validasi perhitungan stok inventory
stock_calculation = (
    inventory["opening_stock"]
    + inventory["received_quantity"]
    - inventory["used_quantity"]
    - inventory["waste_quantity"]
)

stock_valid = np.allclose(
    inventory["closing_stock"],
    stock_calculation,
    atol=0.03
)

# 5. Validasi stok tidak negatif
stock_nonnegative = (
    inventory["closing_stock"] >= 0
).all()

# 6. Validasi durasi table_sessions
session_duration = (
    table_sessions["end_time"] -
    table_sessions["start_time"]
).dt.total_seconds() / 60

duration_valid = np.allclose(
    table_sessions["duration_minutes"],
    session_duration
)

# 7. Validasi jam operasional table_sessions
session_hours_valid = (
    (table_sessions["start_time"].dt.hour >= 10) &
    (
        (table_sessions["end_time"].dt.hour < 22) |
        (
            (table_sessions["end_time"].dt.hour == 22) &
            (table_sessions["end_time"].dt.minute == 0)
        )
    )
).all()

# 8. Validasi feedback
feedback_rating_valid = (
    feedback["rating"].between(1, 5)
).all()

feedback_completed_valid = (
    feedback["order_id"].isin(
        orders.loc[
            orders["status"] == "Completed",
            "order_id"
        ]
    )
).all()

# Menampilkan hasil
checks = {
    "Waktu transaksi dalam jam operasional": order_hours_valid,
    "Subtotal sesuai quantity x unit_price": subtotal_valid,
    "Harga item sesuai harga menu": price_valid,
    "Perhitungan stok inventory benar": stock_valid,
    "Stok akhir tidak negatif": stock_nonnegative,
    "Durasi sesi sesuai waktu mulai dan selesai": duration_valid,
    "Sesi meja dalam jam operasional": session_hours_valid,
    "Rating feedback dalam rentang 1-5": feedback_rating_valid,
    "Feedback berasal dari transaksi Completed": feedback_completed_valid
}

for check, result in checks.items():
    print(
        f"{check}:",
        "Valid" if result else "Perlu diperiksa"
    )
```
